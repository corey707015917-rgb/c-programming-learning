```python

df.isna().sum()
#df.isna() 检查 DataFrame 中的每个元素是不是缺失值。
#缺失值：返回 True。不是缺失值：返回 False。
#.sum() 对每一列的结果进行统计。
# 在求和时，True 被计为 1，False 被计为 0。
#整个表达式统计的是：每一列有多少个缺失值。
```

```python
#复制一份数据，用来练习，保留原始 df
df_clean = df.copy()

#用能耗列的平均值填补缺失能耗
df_clean["energy_kwh"] = df_clean["energy_kwh"].fillna(
   df_clean["energy_kwh"].mean()
)

#用温度列的平均值填补缺失温度
df_clean["temperature_c"] = df_clean["temperature_c"].fillna(
   df_clean["temperature_c"].mean()
)

df_clean
```

```python
df_clean.isna().sum()
#df_clean 是我们自己取的变量名，
# 不是 Python 的特殊命令。 
# 也可以叫 df_filled（填补后的表格），
# 只要后面使用时名字一致就行。

#.isna()：检查每个位置是不是缺失值。
#.sum()：按列加起来。
```
